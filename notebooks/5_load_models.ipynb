# Load Saved Models
This notebook demonstrates how to load the 38 pre-trained LightGBM models from disk.
Run `4_modeling.ipynb` first to generate these `.pkl` files.

```python
import joblib
import os

model_dir = '../output/models/'
loaded_models = {}

if os.path.exists(model_dir):
    for filename in os.listdir(model_dir):
        if filename.endswith('.pkl'):
            # Extract country code (e.g., 'model_DE.pkl' -> 'DE')
            country = filename.replace('model_', '').replace('.pkl', '')
            loaded_models[country] = joblib.load(os.path.join(model_dir, filename))
            
    print(f"{len(loaded_models)} adet model diskten başarıyla geri yüklendi.")
    print("Yüklenen ülkeler:", list(loaded_models.keys()))
else:
    print(f"Hata: {model_dir} klasörü bulunamadı. Lütfen önce 4_modeling.ipynb dosyasını çalıştırın.")
```
